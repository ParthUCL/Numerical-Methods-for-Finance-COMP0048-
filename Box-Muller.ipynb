```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
from scipy.stats import norm 
```

```python
n = np.power(10,6)
x = np.linspace(-4,4,201)
```

```python
U1 = np.random.uniform(0,1,n)
U2 = np.random.uniform(0,1,n)

theta = 2*np.pi*U2
p = np.sqrt(-2*(np.log(U1)))

Z1 = p*np.cos(theta)
Z2 = p*np.sin(theta)
```

```python
plt.figure(1)
plt.plot(x, norm(0,1).pdf(x), label = 'PDF')
plt.plot(x, norm(0,1).cdf(x), label = 'CDF')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Standard Normal Distribution')
plt.legend()
plt.show()
```

```python
plt.figure(2)
plt.hist(Z1, bins=50, density=True, label = 'Sample pdf (Z1)')
plt.hist(Z2, bins=50, density=True, label = 'Sample pdf (Z2)')
plt.plot(x, norm(0,1).pdf(x), label = 'True pdf')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Standard Normal Distribution')
plt.legend()
plt.show()
```
