```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
from scipy.stats import norm 
```

```python
n = np.power(10,6)
x = np.linspace(-4,4,201)
```

```python
plt.figure(1)
plt.plot(x, norm(0,1).pdf(x), label = 'PDF')
plt.plot(x, norm(0,1).cdf(x), label = 'CDF')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Standard Normal Distribution')
plt.legend()
plt.show()
```

```python
Z1 = []
Z2 = []

for i in range(n):
    u1 = np.random.uniform(0,1)
    u2 = np.random.uniform(0,1)
    v1 = 2*u1 - 1
    v2 = 2*u2 - 1
    w = v1**2 + v2**2
    if w < 1:
        Z1.append(v1*np.sqrt(-2*np.log(w) / w))
        Z2. append(v2*np.sqrt(-2*np.log(w) / w))
```

```python
plt.figure(2)
plt.hist(Z1, bins=50, density=True, label = 'Sample pdf (Z1)')
plt.hist(Z2, bins=50, density=True, label = 'Sample pdf (Z2)')
plt.plot(x, norm(0,1).pdf(x), label = 'True pdf')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Standard Normal Distribution')
plt.legend()
plt.show()
```
