```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
npaths = 20000
T = 1
nsteps = 200
dt = T / nsteps
t = np.linspace(0,T,nsteps+1)
sigma = 0.3
b = 1
S0 = 0.8
```

```python
S = np.zeros((npaths, nsteps+1))

for j in range(npaths):
    for i in range(1,nsteps+1):
        S[j][0] = S0
        S[j][i] = S[j][i-1] + ((b - S[j][i-1]) / (T - (i-1)*dt))*dt + sigma*np.sqrt(dt)*np.random.normal(0,1)
```

```python
plt.figure()
plt.plot(t,S.T)
plt
```
