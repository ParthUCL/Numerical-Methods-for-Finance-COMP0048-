```python
import numpy as np
import matplotlib.pyplot as plt 
import scipy 
from scipy.stats import norm, lognorm 
```

```python
mu = 0.2
sigma = 0.1
n = np.power(10,6)
x = np.linspace(0,2,101)
```

```python
plt.figure(1)
plt.plot(x, lognorm(sigma, scale=np.exp(mu)).pdf(x), label='PDF')
plt.plot(x, lognorm(s=sigma, scale=np.exp(mu)).cdf(x), label = 'CDF')
plt.xlabel('x')
plt.title('Lognormal distribution with mu=0.2, sigma=0.1')
plt.legend()
plt.show()
```

```python
U = np.random.uniform(0,1,n)
X = norm(mu,sigma).ppf(U)
Y = np.exp(X)
```

```python
plt.figure(2)
plt.hist(Y, bins=50, density=True, label='Sample pdf')
plt.plot(x, lognorm(s=sigma, scale=np.exp(mu)).pdf(x), label = 'True pdf')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Normal distribution with mu=0.2, sigma=0.1')
plt.legend()
plt.show()
```
