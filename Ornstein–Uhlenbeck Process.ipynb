```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
from scipy.stats import norm
```

```python
npaths = 20000;
T = 1; 
nsteps = 200
dt = T/nsteps
t = np.linspace(0,T,nsteps+1)
theta = 5
mu = 0.07
sigma = 0.07
S0 = 0.03
```

```python
S = np.zeros((npaths, nsteps+1))

for j in range(npaths):
    for i in range(1,nsteps+1):
        S[j][0] = S0
        S[j][i] = S[j][i-1] + theta*(mu - S[j][i-1])*dt + sigma*np.sqrt(dt)*np.random.normal(0,1)
```

```python
plt.figure(1)
plt.plot(t,S.T)
plt.xlabel('t')
plt.title('Ornstein–Uhlenbeck Process  (mu = 0.2, sigma = 0.4, alpha = 5)')
plt.show()
```
