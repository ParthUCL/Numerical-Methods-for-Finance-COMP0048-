```python
import numpy as np
import matplotlib.pyplot as plt
import time
```

```python
npaths = 20000
T = 1
nsteps = 200
dt = T / nsteps 
t = np.linspace(0,1,nsteps+1)
mu = -0.05
sigma = 0.4
```

```python
start = time.process_time()
S = np.zeros((npaths, nsteps + 1))

for j in range(npaths):
    for i in range(1, nsteps + 1):
        S[j][i] = S[j][i-1] + mu*dt + sigma*np.sqrt(dt)*np.random.normal(0,1)
end = time.process_time()

print(end-start)
```
