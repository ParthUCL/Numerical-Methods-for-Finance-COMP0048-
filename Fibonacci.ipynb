```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
from scipy.stats import uniform 
```

```python
i = 17
j = 5
a = 1366
b = 150889
M = 714025
N0 = 12345
n = np.power(10,4)
x = np.linspace(0,1,101)
```

```python
U1 = [N0]
for k in range(1,i):
    U1.append((a*U1[k-1] + b) % M)
```

```python
for l in range(i,n):
    U1.append((U1[l-i] - U1[l-j]) % M)

for m in range(len(U1)):
    U1[m] = U1[m] / M
```

```python
plt.figure(1)
plt.hist(U1, bins = 50, density=True, color = 'blue', edgecolor = 'black', label = 'Sample pdf')
plt.plot(x, uniform(0,1).pdf(x), label = 'True pdf', color='orange')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Uniform Distribution pdf')
plt.legend()
plt.show()
```

```python
U2 = []
for i in range(len(U1)):
    U2.append(U1[i-1])
```

```python
plt.figure(2)
plt.scatter(U1, U2, s=0.5, color = 'blue')
plt.xlabel('U1')
plt.ylabel('U2')
plt.title('2D scatter plot')
plt.show()
```

```python
U3 = []
for i in range(len(U1)):
    U3.append(U1[i-2])
```

```python
fig = plt.figure(3)
ax = fig.add_subplot(projection='3d')
ax.scatter(U1, U2, U3, s=0.5, color = 'blue')
ax.set_xlabel('U1')
ax.set_ylabel('U2')
ax.set_zlabel('U3')
ax.set_title('3D scatter plot')
ax.set_box_aspect(None, zoom=0.80)
plt.show()
```
