```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
npaths = 20
T = 1
nsteps = 200
dt = T / nsteps
t = np.linspace(0,T,nsteps+1)
theta = 5
mu = 0.07
sigma = 0.8
S0 = 0.03
```

```python
S = np.zeros((npaths, nsteps+1))

for j in range(npaths):
    for i in range(1,nsteps+1):
        S[j][0] = S0
        S[j][i] = S[j][i-1] + theta*(mu - S[j][i-1])*dt + sigma*np.sqrt(S[j][i-1])*np.sqrt(dt)*np.random.normal(0,1)
```

```python
plt.figure(1)
plt.plot(t,S.T)
plt.xlabel('t')
plt.title('Feller-Square Root Process  (mu = 0.07, sigma = 0.8, theta = 5)')
plt.show()
```
