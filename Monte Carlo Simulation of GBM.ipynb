```python
import numpy as np
import matplotlib.pyplot as plt
import time
```

```python
npaths = 20000
T = 1
nsteps = 200
dt = T / nsteps 
t = np.linspace(0,1,nsteps+1)
mu = 0.2
sigma = 0.4
muABM = mu - 0.5*(sigma**2)
sigmaABM = sigma
S0 = 1
```

```python
start = time.process_time()
S1 = np.zeros((npaths, nsteps + 1))

for j in range(npaths):
    for i in range(1, nsteps + 1):
        S1[j][i] = S1[j][i-1] + muABM*dt + sigmaABM*np.sqrt(dt)*np.random.normal(0,1)
end = time.process_time()

S2 = S0*np.exp(S1)

print(end-start)
```

```python
plt.figure(1)
plt.plot(t,S2.T)
plt.xlabel('t')
plt.title('Geometric Brownian Motion (mu = 0.2, sigma = 0.4)')
plt.show()
```
