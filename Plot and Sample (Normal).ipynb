```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
from scipy.stats import norm
```

```python
mu = 0.2
sigma = 0.1
n = np.power(10,8)
x = np.linspace(-0.4,0.8,201)
```

```python
plt.figure(1)
plt.plot(x, norm(mu,sigma).pdf(x), label = 'PDF')
plt.plot(x, norm(mu,sigma).cdf(x), label = 'CDF')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Normal Distribution with mu=0.2, sigma=0.1')
plt.legend()
plt.show()
```

```python
U = np.random.uniform(0,1,n)
X = norm(mu,sigma).ppf(U)
```

```python
plt.figure(2)
plt.hist(X, bins=50, density=True, label='Sample pdf')
plt.plot(x, norm(mu,sigma).pdf(x), label = 'PDF')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Normal Distribution with mu=0.2, sigma=0.1')
plt.legend()
plt.show()
```
