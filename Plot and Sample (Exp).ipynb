```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
from scipy.stats import expon
```

```python
tau = 0.7
n = np.power(10,6)
x = np.linspace(0,5,101)
```

```python
plt.figure(1)
plt.plot(x, expon(scale=tau).pdf(x), label = 'PDF')
plt.plot(x, expon(scale=tau).cdf(x), label = 'CDF')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Exponential Distribution with tau=0.7')
plt.legend()
plt.show()
```

```python
U = np.random.uniform(0,1,n)
X = -tau*np.log(1-U)  
```

```python
plt.figure(2)
plt.hist(X, bins=100, density=True, label = 'Sample PDF')
plt.plot(x, expon(scale=tau).pdf(x), label = 'True PDF')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Exponential Distribution with tau=0.7')
plt.legend()
plt.show()
```
