```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
from scipy.stats import norm
```

```python
n = np.power(10,6)
x = np.linspace(-4,4,201)
```

```python
plt.figure(1)
plt.plot(x, norm(0,1).pdf(x), label = 'PDF')
plt.plot(x, norm(0,1).cdf(x), label = 'CDF')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Standard Normal Distribution')
plt.legend()
plt.show()
```

```python
f = norm(0,1).pdf(x)
g = np.exp(-np.abs(x)) / 2
c = np.sqrt(2*np.e / np.pi)
```

```python
Y = []
for i in range(n):
    u1 = np.random.uniform(0,1)
    if u1 < 1/2:
        y = np.log(2*u1)
    if u1 >= 1/2:
        y = -np.log(2-2*u1)
    u2 = np.random.uniform(0,1)
    if u2*c*(np.exp(-np.abs(y)) / 2) <= norm(0,1).pdf(y):
        Y.append(y)    
```

```python
plt.figure(2)
plt.hist(Y, bins=50, density=True, label = 'Sample pdf')
plt.plot(x, norm(0,1).pdf(x), label = 'True pdf')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Standard Normal Distribution')
plt.legend()
plt.show()
```
