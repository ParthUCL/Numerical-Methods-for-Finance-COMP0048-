```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
T = 1
X = 1.1
```

```python
r = 0.2
q = 0.02
S0 = 1
```

```python
sigma = 0.4
muABM = r - q - 0.5*(sigma**2)
sigmaABM = sigma
```

```python
nblocks = 10000
npaths = 2000
```

```python
VcMCb = []
VpMCb = []

for t in range(nblocks):
    S1 = np.zeros((npaths, 1))

    for j in range(npaths):
        S1[j][0] = muABM*T + sigmaABM*np.sqrt(T)*np.random.normal(0,1)
        S2 = S0*np.exp(S1)

    VcMCb.append(np.exp(-r*T)*np.mean(np.maximum(S2-X,0)))
    VpMCb.append(np.exp(-r*T)*np.mean(np.maximum(X-S2,0))) 
```

```python
VcMC = np.mean(VcMCb)
VpMC = np.mean(VpMCb)

scMC = np.sqrt(np.var(VcMCb)/nblocks)
spMC = np.sqrt(np.var(VpMCb)/nblocks)

print(VcMC)
print(VpMC)
```
