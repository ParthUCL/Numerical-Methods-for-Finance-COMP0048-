```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
from scipy.stats import ncx2
```

```python
k = 5
l = 2
n = np.power(10,6)
x = np.linspace(0,25,101)
```

```python
plt.figure(1)
plt.plot(x, ncx2(k,l).pdf(x), label = 'PDF')
plt.plot(x, ncx2(k,l).cdf(x), label = 'CDF')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('NCX2 Distribution with k = 5, l = 2')
plt.legend()
plt.show()
```

```python
U = np.random.uniform(0,1,n)
X = ncx2(k,l).ppf(U)
```

```python
plt.figure(2)
plt.hist(X, bins=50, density=True, label = 'Sample pdf')
plt.plot(x, ncx2(k,l).pdf(x), label = 'True pdf')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('NCX2 Distribution with k = 5, l = 2')
plt.legend()
plt.show()
```
